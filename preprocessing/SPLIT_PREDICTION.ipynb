```python
import os
from pathlib import Path
import nibabel as nib
import numpy as np
```

```python
def split_arrays(original):
    nptype = original.dtype
    # First array: 1 for values around 1, 0 otherwise
    array1 = np.where(np.isclose(original, 1, atol=0.01), 1, 0)
    
    # Second array: 1 for values around 2, 0 otherwise
    array2 = np.where(np.isclose(original, 2, atol=0.01), 1, 0)
    
    return array1.astype(nptype), array2.astype(nptype)
```

```python
PRED_PATH = '/data_network/summer_projects/geninana/Current/2020_AINI-Stroke_TC_BL/2025_AVA_Stroke/IRM/OUTPUTS'
```

```python
for pred_file in os.listdir(PRED_PATH):
    if pred_file.endswith('.nii.gz'):
        pred_path = os.path.join(PRED_PATH, pred_file)
        if '_pred.' in pred_file:
            print("PRED:", pred_path)
            brain_mask_path = pred_path.replace('_pred.', '_mask.')
            vessel_mask_path = pred_path.replace('_pred.', '_vessel.')
            
            if os.path.exists(brain_mask_path) and os.path.exists(vessel_mask_path):
                print("Already done")
                continue

            segmentation = nib.load(pred_path)
            segmentation_data = segmentation.get_fdata()
            brain_mask, vessel_mask = split_arrays(segmentation_data)
            brain_mask = nib.Nifti1Image(brain_mask, segmentation.affine)
            vessel_mask = nib.Nifti1Image(vessel_mask, segmentation.affine)


            nib.save(brain_mask, brain_mask_path)
            nib.save(vessel_mask, vessel_mask_path)

        if '_trans.' in pred_file:
            print("TRANS", pred_path)
```

```python
DEST_MASK = '/data_network/summer_projects/geninana/Current/2020_AINI-Stroke_TC_BL/2025_AVA_Stroke/IRM/TOF_nifti_TEST3/raw_mask'
DEST_VESSEL = '/data_network/summer_projects/geninana/Current/2020_AINI-Stroke_TC_BL/2025_AVA_Stroke/IRM/TOF_nifti_TEST3/raw_vessel'
DEST_RAW = '/data_network/summer_projects/geninana/Current/2020_AINI-Stroke_TC_BL/2025_AVA_Stroke/IRM/TOF_nifti_TEST3/raw'

SOURCE_RAW = '/data_network/summer_projects/geninana/Current/2020_AINI-Stroke_TC_BL/2025_AVA_Stroke/IRM/TOF_nifti/'
```

```python
for pred_file in os.listdir(PRED_PATH):
    if pred_file.endswith('.nii.gz'):
        pred_path = os.path.join(PRED_PATH, pred_file)
        if '_mask.' in pred_file:
            print("MASK:", pred_path)
            # MOVE TO DESTINATION
            dest_path = os.path.join(DEST_MASK, pred_file)
            os.rename(pred_path, dest_path)
        if '_vessel.' in pred_file:
            print("VESSEL:", pred_path)
            # MOVE TO DESTINATION
            dest_path = os.path.join(DEST_VESSEL, pred_file)
            os.rename(pred_path, dest_path)
        if '_trans.' in pred_file:
            RAW_FILE = os.path.basename(pred_file).replace('_trans.', '.')
            RAW = os.path.join(SOURCE_RAW, RAW_FILE)
            print("RAW:", os.path.join(RAW))
            assert os.path.exists(RAW)
            # COPY TO DESTINATION
            dest_path = os.path.join(DEST_RAW, RAW_FILE)
            # COPY
            print(f'cp {RAW} {dest_path}')
            os.system(f'cp {RAW} {dest_path}')
            
```
